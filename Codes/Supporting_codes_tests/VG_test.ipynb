For the glacial till trees: Morainal Deposits
Morainal deposits, often referred to as glacial till, are materials laid down by the glacial ice. These deposits are generally comprised of stones and gravels embedded in a matrix of sand, silt and clay sized materials. When this matrix contains nearly equal amounts of sand, silt and clay they are called loamy morainal deposits. When there are larger amounts of sand, they are referred to as sandy morainal. Source:  Saskatchewan Land Resource Unit, 2009. Saskatchewan Soil Resource Database User's Manual for SKSIDv4. Agriculture and Agri-Food Canada, Saskatoon, Sask.(https://soilsofsask.ca/soil-survey-soil-characteristics/surface-deposits.php#:~:text=Morainal%20deposits%2C%20often%20referred%20to,silt%20and%20clay%20sized%20materials.)

Soil clay contents range from 0.025 to 0.375 kg kg−1 (Table 1). The graded soil texture of the glacial tills, generally with a modest content of clay in combination with (coarse) sand, tends to induce very high densities. This is especially the case in the subsoil, where a bulk density in the range 1.6–1.8 g cm3 is not uncommon. The low content of clay implies that the soils included are poorly structured.

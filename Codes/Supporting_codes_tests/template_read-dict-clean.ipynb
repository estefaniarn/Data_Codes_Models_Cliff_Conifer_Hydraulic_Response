```python
from pathlib import Path

from csv import DictReader, reader
# Import the utility functions
from utils.read_and_create_dict import read_and_create_dict
from utils.clean_psy import clean_psy_in_dict

# Read files, create/update the dictionary, and clean PSY data
trees_data, paths = read_and_create_dict()
trees_data = clean_psy_in_dict(trees_data)
```

```python
# plot all tree sf for each year
years = ['2021','2022']

for year in years:
    fig = go.Figure()
    for data in trees_data.values():
        # Only include trees from the current year and with valid SF data
        if data['year'] == year and data['SF'] is not None and not data['SF'].empty:
            fig.add_trace(go.Scatter(
                x=data['SF'].index,
                y=data['SF']['SF_incomplete'],  
                mode='lines',
                name=data['tree']
            ))

# Customize the layout
    fig.update_layout(
        title=f'Sapflow Series for {year}',
        xaxis_title='Date',
        yaxis_title='Sapflow (cm3/h)',
        template='plotly_white',
    )
    fig.show()

```

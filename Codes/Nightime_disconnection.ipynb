## Predawn regression 

```python
from utils.predawn import calculate_predawn_water_potential, plot_predawn_scatter

# Calculate predawn water potentials (Ψ_pd) and store in trees_data
trees_data = calculate_predawn_water_potential(trees_data)

# Plot nighttime scatter plots for 2021 and 2022 Sap Flow vs. Water Potential (10 PM to 7 AM) for {tree_name} in {year}, Colored by Day
plots_pd_2021 = plot_predawn_scatter(trees_data, '2021')
plots_pd_2022 = plot_predawn_scatter(trees_data, '2022')

# Path
path_predawn = path_graphs + 'Predawn_WP_SSplots/'

# Save 2021 plots (PNG,HTML)
for fig, tree_name in plots_pd_2021:
    if fig is not None:  # Ensure the figure is valid
        fig.write_image(path_predawn + f'predawn_potential_2021_{tree_name}_2021.png', scale=8)
        fig.write_html(path_predawn + f'predawn_potential_2021_{tree_name}_2021.html', include_plotlyjs='cdn')
        # fig.show() #Uncomment to display 2021 plots here

# Save 2022 plots (PNG,HTML)
for fig, tree_name in plots_pd_2022:
    if fig is not None:  # Ensure the figure is valid
        fig.write_image(path_predawn+ f'predawn_potential_2022_{tree_name}_2022.png', scale=8)
        fig.write_html(path_predawn + f'predawn_potential_2022_{tree_name}_2022.html', include_plotlyjs='cdn')
        # fig.show() #Uncomment to display 2022 plots here

```

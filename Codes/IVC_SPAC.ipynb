```python
# Define trees to exclude (default: DF49-2021, DF03-2022, can be extended)
exclude_trees = [
    ('DF49', 2021),
    ('DF03', 2022),
    ('ES50', 2022),
#    ('DF27', 2021),
]
# Add more tuples (tree, year) to exclude_trees if needed, e.g., exclude_trees.append(('DF01', 2022))

# Load the lumped results
df = pd.read_csv('lumped_weibull_fits.csv')

# Filter out excluded trees
for tree, year in exclude_trees:
    df = df[~((df['Tree'] == tree) & (df['Year'] == year))]

# Weibull function
def weibull_func(x, r_min, d, b):
    return r_min * np.exp(((-x / d) ** b))

# Create figure
fig = go.Figure()

# Plot each Weibull fit
for _, row in df.iterrows():
    if not any(np.isnan([row['Weibull_Rmin'], row['Weibull_d'], row['Weibull_b']])):
        x_fit = np.linspace(-10, 0, 100)  # Range from -10 to 0 MPa
        y_fit = weibull_func(x_fit, row['Weibull_Rmin'], row['Weibull_d'], row['Weibull_b'])
        fig.add_trace(go.Scatter(
            x=x_fit,
            y=y_fit,
            mode='lines',
            name=f"{row['Tree']}-{row['Year']}",
            line=dict(width=2)
        ))

# Update layout
fig.update_layout(
    height=400,
    width=600,
    showlegend=True,
    legend=dict(x=1.02, y=1.0, xanchor='left', yanchor='top'),
    margin=dict(r=50, t=50, b=50, l=50),
    xaxis_title="PSYx (MPa)",
    xaxis_range=[-2, 0],
    yaxis_title="Rp (MPa s kg⁻¹)",
    yaxis_range=[0, 40000] #60000
)

fig.show()

# # Save plot
# fig.write_image(path_resistance + 'R_ALL_Weibull.png', scale=9)
# fig.write_html(path_resistance + 'R_ALL_Weibull.html', include_plotlyjs='cdn')

```

```python
# Define trees to exclude (default: DF49-2021, DF03-2022, ES50-2022, DF27-2021, can be extended)
exclude_trees = [
    ('DF49', 2021),
    ('DF03', 2022),
    ('ES50', 2022),
#    ('DF27', 2021),
]
# Add more tuples (tree, year) to exclude_trees if needed, e.g., exclude_trees.append(('DF01', 2022))

# Load the lumped results
df = pd.read_csv('lumped_weibull_fits.csv')

# Filter out excluded trees
for tree, year in exclude_trees:
    df = df[~((df['Tree'] == tree) & (df['Year'] == year))]

# Weibull function
def weibull_func(x, r_min, d, b):
    return r_min * np.exp(((-x / d) ** b))

# Create figure
fig = go.Figure()

# Plot each Weibull fit for Kp (1/Rp)
for _, row in df.iterrows():
    if not any(np.isnan([row['Weibull_Rmin'], row['Weibull_d'], row['Weibull_b']])):
        x_fit = np.linspace(-10, 0, 100)  # Range from -10 to 0 MPa
        rp_fit = weibull_func(x_fit, row['Weibull_Rmin'], row['Weibull_d'], row['Weibull_b'])
        kp_fit = 1 / rp_fit  # Compute Kp as inverse of Rp
        fig.add_trace(go.Scatter(
            x=x_fit,
            y=kp_fit,
            mode='lines',
            name=f"{row['Tree']}-{row['Year']}",
            line=dict(width=2)
        ))

# Update layout
fig.update_layout(
    height=400,
    width=600,
    showlegend=True,
    legend=dict(x=1.02, y=1.0, xanchor='left', yanchor='top'),
    margin=dict(r=50, t=50, b=50, l=50),
    xaxis_title="PSYx (MPa)",
    xaxis_range=[-6, 0],
    yaxis_title="Kp (kg MPa⁻¹ s⁻¹)",
    yaxis_range=[0, 0.00015],  # 0 to 0.0003 for cliff trees, 0 to 0.008 for soil trees
    yaxis=dict(
        tickformat="f",  # Use plain integer format without "μ"
        ticksuffix=" "     # Remove any suffix like "μ"
    )
)

fig.show()



# # Save plot when range is 0-0.015
# fig.write_image(path_resistance + 'Conductance_big.png', scale=9) 
# fig.write_html(path_resistance + 'Conductance_big.html', include_plotlyjs='cdn')

# # Save plot when range is 0-0.0008
# fig.write_image(path_resistance + 'Conductance_small.png', scale=9) 
# fig.write_html(path_resistance + 'Conductance_small.html', include_plotlyjs='cdn')
```

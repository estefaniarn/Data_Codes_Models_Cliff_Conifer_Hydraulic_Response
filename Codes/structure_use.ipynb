## Radiation

```python
import plotly.graph_objects as go
import pandas as pd

# plot longest tree sf for each year with additional calculations and aligned date ranges
years = ['2021','2022']

for year in years:
    year_datas = [
        data for data in trees_data.values() 
        if data['year'] == year 
        and 'SF' in data 
        and data['SF'] is not None 
        and not data['SF'].empty
        and data['SF'].index.min().month <= 5
        and data['SF'].index.max().month >= 10
    ]
    
    if not year_datas:
        continue
    
    longest_data = max(year_datas, key=lambda d: len(d['SF']))
    sf_df = longest_data['SF'].copy()
    
    # Align date ranges: May 5 to Oct 19 for both years
    start_aligned = pd.to_datetime(f"{year}-05-06")
    end_aligned = pd.to_datetime(f"{year}-10-19 23:59:59")
    sf_df = sf_df[(sf_df.index >= start_aligned) & (sf_df.index <= end_aligned)]
    
    if sf_df.empty:
        print(f"No data in aligned range for {year}")
        continue
    
    # Calculations on aligned data
    sf_df['date'] = sf_df.index.date
    sf_df['hour'] = sf_df.index.hour
    
    midday_df = sf_df[(sf_df['hour'] >= 11) & (sf_df['hour'] <= 14)]
    daily_midday = midday_df.groupby('date')['NR'].mean().reset_index()
    
    total_days = len(daily_midday)
    high_days = daily_midday[daily_midday['NR'] > 300]
    num_high = len(high_days)
    percentage = (num_high / total_days) * 100 if total_days > 0 else 0
    avg_high = high_days['NR'].mean() if num_high > 0 else 0
    
    print(f"For {year} ({longest_data['tree']}):")
    print(f"Total days with 30-min measurements (aligned): {total_days}")
    print(f"Days with midday NR > 300: {num_high}")
    print(f"Percentage: {percentage:.2f}%")
    
    fig = go.Figure()
    fig.add_trace(go.Scatter(
        x=sf_df.index,
        y=sf_df['NR'],  
        mode='lines',
        name=longest_data['tree']
    ))
    
    # Overall average midday
    avg_midday = daily_midday['NR'].mean()
    fig.add_hline(
        y=avg_midday, 
        line_dash="dash", 
        annotation_text=f"Avg Midday: {avg_midday:.2f}", 
        annotation_position="bottom right"
    )
    
    # If percentage >= 40%, add average of high days
    if percentage >= 40:
        fig.add_hline(
            y=avg_high, 
            line_dash="dot", 
            line_color="red",
            annotation_text=f"Avg High Midday (>300): {avg_high:.2f}", 
            annotation_position="top right"
        )
    
    # Set x-axis to aligned range
    fig.update_xaxes(range=[start_aligned, end_aligned])
    
    # Customize the layout
    fig.update_layout(
        #title=f'Net Radiation {year} ({longest_data["tree"]}) | N = {total_days} days',
        xaxis_title='Date',
        yaxis_title='Net Radiation (W m^-2)',
        template='plotly_white',
    )
    fig.show()
    fig.write_image(path_time_series+f'NR_{year}.png',scale=10)
    fig.write_html(path_time_series + f'NR_{year}.html', include_plotlyjs='cdn')


```
